```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))
```

```python
sigmoid(1)
```

```python
sigmoid(1000)
```

```python
def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (math.exp(-x) + math.exp(x))
```

```python
tanh(-50)
```

```python
tanh(50)
```
