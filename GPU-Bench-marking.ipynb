```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
from tensorflow import keras
```

```python
tf.config.experimental.list_physical_devices()
```

```python
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.cifar10.load_data()
```

```python
X_train.shape # we will have 3 becase is red, green, blue
```

```python
def plot_image(index):
    plt.figure(figsize = (10, 1))
    plt.imshow(X_train[index])
```

```python
plot_image(0)
```

```python
classes = ["airplane", "automobile", "bird", "cat", "deer", "dog", "frog", "horse", "ship", "truck"]
```

```python
classes[y_train[0][0]]
```

```python
X_train_scaled = X_train / 255
X_test_scaled = X_test / 255
```

```python
y_train[:5]
```

```python
y_train_categorical = keras.utils.categorical(
    y_train, num_classes = 10, dtype = 'float32'
)
```

```python
y_test_categorical = keras.utils.categorical(
    y_test, num_classes = 10, dtype = 'float32'
)
```

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape = (32, 32, 3)),
    keras.layers.Dense(50000, activation = 'relu')
    keras.layers.Dense(10000, activation = 'relu')
    keras.layers.Dense(10, activation = 'sigmoid')
])
```

```python
model.compile(
    optimizer = 'adam'
    loss = 'categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
model.fit(X_train_scaled, y_train_categorical, epochs = 5)
```

```python
model.evaluate(X_test_scaled, y_test_scaled)
```

```python
classes[np.argmax(model.predict(X_test_scaled)[0])]
```

```python
def get_model():
    model = keras.Sequential([
         keras.layers.Flatten(input_shape = (32, 32, 3)),
         keras.layers.Dense(50000, activation = 'relu')
         keras.layers.Dense(10000, activation = 'relu')
         keras.layers.Dense(10, activation = 'sigmoid')
    ])
    model.compile(
        optimizer = 'adam'
        loss = 'categorical_crossentropy',
        metrics = ['accuracy']
   )

   return model
```

```python
%%timeit -n1 -r1
with tf.device('/CPU:0'):
    cpu_model = get_model()
    cpu_model.fit(X_train_scaled, y_train_categorical, epochs= 1)
```

```python
%%timeit -n1 -r1
with tf.device('/GPU:0'): # I dont have it already
    gpu_model = get_model()
    gpu_model.fit(X_train_scaled, y_train_categorical, epochs = 1)
```
