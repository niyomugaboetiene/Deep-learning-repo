```python
import tensorflow as tf
```

```python
import matplotlib.pyplot as plt
import numpy as np
from tensorflow import keras
%matplotlib inline
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data() # this is how we load datasets in tensoflow
# mnist is handrwitten dataset with huge collection of numbers
# there are also Fashion mnist
# CIFAR-10 for cars, cats, airplanes, ...
```

```python
# We need to scale the data for better training
X_train = X_train / 255
X_test = X_test / 255
```

```python
X_train.shape
# We need to convert the 2D Into 1D
(60000, 784)
```

```python
X_train_flattened = X_train.reshape(len(X_train), 28 * 28)
X_test_flattened = X_test.reshape(len(X_test), 28 * 28)
X_train_flattened.shape
```

```python
X_test_flattened.shape
```

```python
print(X_train_flattened.shape)
print(y_train.shape)

print(X_test_flattened.shape)
print(y_test.shape)
```

```python
X_train_flattened[0]
```

```python
model = keras.Sequential([
    keras.layers.Dense(10, input_shape = (784,), activation='sigmoid')
])
# .Dense means every input layer will connected to each output layer and define the input numbers and the output which is 10
# activation='sigmoid' will convert number into range of 0 and 1

# define how I want my model to train
model.compile( 
    optimizer='adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
model.fit(X_train_flattened, y_train, epochs = 9) # it will repeats 9 times
```

```python
# test the model
model.evaluate(X_test_flattened, y_test) 
```

```python
plt.imshow(X_test[0])
```

```python
predicted = model.predict(X_test_flattened)
```

```python
predicted[0]
```

```python
np.argmax(predicted[0])
```

```python
y_predicted_labels = [np.argmax(i) for i in predicted]

y_predicted_labels[:4]
```

```python
y_test[:4]
```

```python
cm = tf.math.confusion_matrix(labels=y_test, predictions = y_predicted_labels)
cm
```

```python
import seaborn as sns

plt.figure(figsize= (10, 7))
sns.heatmap(cm, annot = True, fmt = 'd')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
model = keras.Sequential([
    keras.layers.Dense(100, input_shape = (784,), activation='relu'),
    keras.layers.Dense(10, activation='sigmoid')
])
# .Dense means every input layer will connected to each output layer and define the input numbers and the output which is 10
# activation='sigmoid' will convert number into range of 0 and 1

# define how I want my model to train
model.compile( 
    optimizer='adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
model.fit(X_train_flattened, y_train, epochs = 9) # it will repeats 9 times
```

```python
# test the model
model.evaluate(X_test_flattened, y_test) 
```

```python
predicted = model.predict(X_test_flattened)
y_predicted_labels = [np.argmax(i) for i in predicted]
cm = tf.math.confusion_matrix(labels=y_test, predictions = y_predicted_labels)

plt.figure(figsize= (10, 7))
sns.heatmap(cm, annot = True, fmt = 'd')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
model.export("./saved_model")
```

```python
# methood 1 Post training
```

```python
converter = tf.lite.TFLiteConverter.from_saved_model("./saved_model") # this one can also make model small but with quantization make it more smaller
tflite_model = converter.convert()
```

```python
len(tflite_model)
```

```python
# with quantization
converter = tf.lite.TFLiteConverter.from_saved_model("./saved_model")
converter.optimizations = [tf.lite.Optimize.DEFAULT]
quantized_model = converter.convert()
```

```python
len(quantized_model) # now it is reduced
```

```python
# save now quantizrd model
with open("tflite_model.tflite", "wb") as f:
    f.write(tflite_model)

with open("quantized_model.tflite", "wb") as f:
    f.write(quantized_model)
```
