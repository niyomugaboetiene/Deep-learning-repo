```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('sonar_dataset.csv', header=None)
df.head()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df[60].value_counts()
```

```python
X = df.drop(60, axis='columns')
y = df[60]
```

```python
y.sample
```

```python
y = pd.get_dummies(df[60], drop_first=True).astype(int)
```

```python
y.sample
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=1)
```

```python
X_train.shape, X_test.shape
```

```python
import tensorflow as tf
from tensorflow import keras
```

```python
model = keras.Sequential([
    keras.layers.Dense(60, input_shape = (60,), activation='relu'),
    keras.layers.Dense(30, activation='relu'),
    keras.layers.Dense(15, activation='relu'),
    keras.layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(X_train, y_train, epochs=100, batch_size=8) # for each iteration it will take 8 sample and measure the accuracy
```

```python
model.evaluate(X_test, y_test)
```

```python
y_pred = model.predict(X_test).reshape(-1)
print(y_pred)
```

```python
# round them all
y_pred = np.round(y_pred)
y_pred[:10]
```

```python
y_test[:10]
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test, y_pred))
```

```python
model2 = keras.Sequential([
    keras.layers.Dense(60, input_shape = (60,), activation='relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(30, activation='relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(15, activation='relu'),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(1, activation='sigmoid')
])
```

```python
model2.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model2.fit(X_train, y_train, epochs=50, batch_size=8) # for each iteration it will take 8 sample and measure the accuracy
```

```python
model2.evaluate(X_test, y_test)
```

```python
y_pred = model2.predict(X_test).reshape(-1)
y_pred = np.round(y_pred)

print(classification_report(y_test, y_pred))
```
