```python
import gensim
import pandas as pd
```

```python
df = pd.read_json('./reviews_Cell_Phones_and_Accessories_5/Cell_Phones_and_Accessories_5.json', lines=True)
```

```python
df.head()
```

```python
df.reviewText[0]
```

```python
gensim.utils.simple_preprocess("They look good and stick good! I just don't like the rounded shape because I was always bumping it and Siri kept popping up and it was irritating. I just won't buy a product like this again")
```

```python
# applying preprocessing to remove puntuations, and other unnecessary stuff

review_text = df.reviewText.apply(gensim.utils.simple_preprocess)
review_text
```

```python
model = gensim.models.Word2Vec(
    window = 10, # how many windows which means 10 words before and after target
    workers = 4, # is how many CPU
    min_count = 2 # force to use word wich has two words if it is only 1 word don't use it
)
```

```python
# building vocal which means unique list set of words
model.build_vocab(review_text, progress_per=1000)
```

```python
model.epochs
```

```python
model.corpus_count
```

```python
# train our model
model.train(review_text, total_examples=model.corpus_count, epochs=model.epochs)
```
