```python
import numpy as np
```

```python
y_pred = np.array([1, 1, 0, 0, 1])
y_true = np.array([0.30, 0.7, 1, 0, 0.5])
```

```python
def mae(y_true, y_pred):
    total_abs = 0
    for yt, yp in zip(y_true, y_pred):
        total_abs += abs(yt - yp)
    print("Total error", total_abs)
    mae = total_abs / len(y_pred)
    print("MAE", mae)
```

```python
mae(y_true, y_pred)
```

```python
np.sum(np.abs(y_true - y_pred))
```

```python
# MSE = sum of (y-y_pred)**2
def mse(y_true, y_pred):
    total_se = 0
    for yt, yp in zip(y_true, y_pred):
        total_se += pow(yt - yp, 2)
    print("Total SE", total_se)

    mse = (total_se / len(y_true))
    print("MSE", mse)
```

```python
mse(y_true, y_pred)
```
