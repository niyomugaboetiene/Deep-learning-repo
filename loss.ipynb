```python
import numpy as np
```

```python
y_pred = np.array([1, 1, 0, 0, 1])
y_true = np.array([0.30, 0.7, 1, 0, 0.5])
```

```python
def mae(y_true, y_pred):
    total_abs = 0
    for yt, yp in zip(y_true, y_pred):
        total_abs += abs(yt - yp)
    print("Total error", total_abs)
    mae = total_abs / len(y_pred)
    print("MAE", mae)
```

```python
mae(y_true, y_pred)
```

```python
np.sum(np.abs(y_true - y_pred))
```

```python
 eplison = 1e-15
```

```python
y_predicted_new = [max(i, eplison) for i in y_pred]
y_predicted_new
```

```python
y_predicted_new = [min(i, 1-eplison) for i in y_predicted_new]
y_predicted_new
```

```python
y_predicted_new = np.array(y_predicted_new)
np.log(y_predicted_new)
```
